```python
# 1.2: Figures 1 & 2 Data Calculations

# --- FIGURE 1 LOGIC ---
# Assuming 'New Profile' represents the Reconnect stage
reconnect_df = intake[intake['Issue Type'] == 'New Profile']

# Define "Eviction Risk" as currently being evicted, unhoused or unknown
urgent_mask = (reconnect_df['Custom field (Eviction Status)'] == 'Currently being evicted') | (reconnect_df['Custom field (Housing Status)'] == 'Unhoused') | (reconnect_df['Custom field (Eviction Status)'] == 'Unknown')


fig1_eviction_risk = reconnect_df[urgent_mask].shape[0]
fig1_housing_stable = reconnect_df[~urgent_mask].shape[0]

print("--- FIGURE 1 RESULTS ---")
print(f"Total Reconnect Tickets: {len(reconnect_df)}")
print(f"Housing Stable: {fig1_housing_stable}")
print(f"Eviction Risk: {fig1_eviction_risk}")

# --- FIGURE 2 LOGIC ---

# Map Frequency (Episodic = 1, Chronic = 2+)
ticket_counts = intake['PID'].value_counts()
intake['Frequency'] = intake['PID'].map(lambda x: 'Episodic' if ticket_counts[x] == 1 else 'Chronic')

# Map Urgency
intake['Urgency'] = np.where(
    (intake['Custom field (Eviction Status)'] == 'Currently being evicted') |
    (intake['Custom field (Housing Status)'] == 'Unhoused'),
    'Urgent', 'Less Urgent'
)

# Displaying the four quadrants
fig2_matrix = pd.crosstab(intake['Frequency'], intake['Urgency'])
print("\n--- FIGURE 2 RESULTS (Quadrants) ---")
display(fig2_matrix)
```

### Q4: Repeat Contact
**Answer: 150 unique individuals submitted multiple tickets. For the vast majority, their immediate crisis state (Eviction/Housing status) did not change between their first and most recent ticket, indicating prolonged instability.**

```python
# Q4 Code
ticket_counts = intake['PID'].value_counts()
repeat_pids = ticket_counts[ticket_counts > 1].index
#Comparing first and last eviction status for each PID
repeats_df = intake[intake['PID'].isin(repeat_pids)].sort_values(by=['PID', 'Created_DT'])
first_tickets = repeats_df.groupby('PID').first()[['Custom field (Housing Status)', 'Custom field (Eviction Status)']]
last_tickets = repeats_df.groupby('PID').last()[['Custom field (Housing Status)', 'Custom field (Eviction Status)']]

comparison_df = first_tickets.join(last_tickets, lsuffix='_First', rsuffix='_Last')
eviction_changed = (comparison_df['Custom field (Eviction Status)_First'] != comparison_df['Custom field (Eviction Status)_Last']).sum()

print(f"Out of {len(repeat_pids)} repeat contacts:")
print(f"Eviction Status changed for only {eviction_changed} people between their first and last ticket.")
```

```python
# 1. & 2. FILTERING SUB-TICKETS & DATA CLEANING LOG

# Sub-tickets are identified by having a non-null 'Parent' or 'Parent key'. We only want Intake Profiles.
intake_profiles = intake[intake['Parent'].isna()].copy()
print(f"Filtered to {len(intake_profiles)} intake profiles (removed {len(intake) - len(intake_profiles)} sub-tickets).\n")

cleaning_log = []

# Handle Income Outliers: Column is already numeric. We remove extreme outliers (> $10,000/mo)
high_income_mask = intake_profiles['Income_Amount'] > 10000
cleaning_log.append(f"Set {high_income_mask.sum()} income outliers (> $10,000) to NaN.")
intake_profiles.loc[high_income_mask, 'Income_Amount'] = np.nan

# Handle Age Outliers: Impossible ages (e.g., > 100)
invalid_age_mask = intake_profiles['Age at Intake'] > 100
cleaning_log.append(f"Set {invalid_age_mask.sum()} invalid ages (> 100) to NaN.")
intake_profiles.loc[invalid_age_mask, 'Age at Intake'] = np.nan

# Handle Employment Missingness on INTAKE PROFILES ONLY (ignores the 213 sub-tickets)
missing_emp = intake_profiles['Custom field (Employment Status)'].isna().sum()
cleaning_log.append(f"Filled {missing_emp} missing Employment Status values in intake profiles with 'Unknown'.")
intake_profiles['Employment_Status'] = intake_profiles['Custom field (Employment Status)'].fillna('Unknown')

print("--- Data Cleaning Log ---")
for log in cleaning_log:
    print("- " + log)
```

```python
# 3. REFACTORED FIGURE 1 & 2

# --- FIGURE 1 ---
# Eviction Risk = 'Currently being evicted' or 'Unhoused'. 'Unknown' is NOT an active risk.
urgent_mask_fixed = (intake_profiles['Custom field (Eviction Status)'] == 'Currently being evicted') | (intake_profiles['Custom field (Housing Status)'] == 'Unhoused')

fig1_eviction_risk = intake_profiles[urgent_mask_fixed].shape[0]
fig1_housing_stable = intake_profiles[~urgent_mask_fixed].shape[0]

print("--- FIGURE 1 RESULTS (Fixed) ---")
print(f"Total Intake Profiles: {len(intake_profiles)}")
print(f"Eviction Risk / Unhoused: {fig1_eviction_risk}")
print(f"Housing Stable / Unknown: {fig1_housing_stable}\n")

# --- FIGURE 2 ---
# Visualizing the Quadrant Matrix instead of just a table
intake_profiles['Frequency'] = intake_profiles['PID'].map(lambda x: 'Episodic' if ticket_counts[x] == 1 else 'Chronic')
intake_profiles['Urgency'] = np.where(urgent_mask_fixed, 'Urgent', 'Less Urgent')

fig2_matrix_fixed = pd.crosstab(intake_profiles['Frequency'], intake_profiles['Urgency'])

plt.figure(figsize=(6, 4))
sns.heatmap(fig2_matrix_fixed, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('Figure 2: Urgency vs. Frequency Quadrants')
plt.xlabel('Crisis Urgency')
plt.ylabel('Contact Frequency')
plt.show()
```

```python
# 4. REFACTORED Q1 (Rates & Household Size by Zip)

# Overall Share
total_profiles = len(intake_profiles)
active_evictions = intake_profiles[intake_profiles['Custom field (Eviction Status)'] == 'Currently being evicted']
overall_share = (len(active_evictions) / total_profiles) * 100
print(f"Overall share of intake profiles facing active eviction: {overall_share:.1f}% ({len(active_evictions)}/{total_profiles})\n")

# Household Size Calculation
# Get count of additional members per PID
hh_size = house.groupby('PID')['Member ID'].count().reset_index(name='Additional_Members')
intake_hh = pd.merge(intake_profiles, hh_size, on='PID', how='left')
# Total size = additional members + 1 (the applicant)
intake_hh['Total_Household_Size'] = intake_hh['Additional_Members'].fillna(0) + 1

# Zip Code Eviction Stats
zip_evictions = intake_hh[intake_hh['Custom field (Eviction Status)'] == 'Currently being evicted']
zip_stats = zip_evictions.groupby('Custom field (Zip Code)').agg(
    Eviction_Count=('PID', 'count'),
    Avg_Household_Size=('Total_Household_Size', 'mean')
).sort_values('Eviction_Count', ascending=False).head(5)

# Add rate column
zip_stats['%_of_Total_Evictions'] = (zip_stats['Eviction_Count'] / len(active_evictions) * 100).round(1).astype(str) + '%'

print("Top 5 Zip Codes for Active Evictions (Counts, Rates, & Avg Household Size):")
display(zip_stats.round(2))
```
